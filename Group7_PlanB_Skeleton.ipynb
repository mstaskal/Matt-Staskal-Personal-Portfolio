# Group 7: NFL Ad Package Optimization (Plan B)

A CBS affiliate sells season-long NFL ad packages: one 30-second ad in each of 27 games.
1. Before the NFL schedule is known, the station picks which prospects to pitch and at what discount.
2. After the schedule is known, each salesperson splits limited hours between fixing
   ratings shortfalls and selling the remaining packages.

Goal: make as much money as possible while owing as few makegoods as possible.
These goals conflict, and this notebook shows the trade-off.

## Plan B at a glance
- **Role:** Backup. Ed's revised plan has the station make one guarantee decision, which Plan A
  models directly. Plan B gets there more simply: it runs the model once per guarantee level and compares.
- **Linear?** Yes. All the math with "set to 0 if negative" happens in pandas, before Gurobi.
- **Course module:** Module 3 (two stages, scenarios, two objectives). Module 4 only to declare
  yes/no and whole-number variables.
- **Size:** 126 decision variables and 70 constraints across 5 families. The rubric needs 50, 50, and 3.
- **Use it if** Plan A is too complicated.

## Decision for Sunday: one model or two?
Ed's plan says the station can't see the schedule in July. So Stage 1 should judge risk with the
July view (`shortfall_stage1`), and Stage 2 should react to the real schedule (`shortfall_stage2`).

- **One model (as written below):** Both stages are solved together, and makegood cost uses
  `exp_mg_stage2`. It's simpler and bigger for the rubric, but Stage 1 effectively "knows" the dud
  games in July. We'd state that as a limitation in the report.
- **Two models in sequence:** Model 1 picks pitches using `exp_mg_stage1`. Those pitches are then
  locked in, and Model 2 decides sell and fix using `exp_mg_stage2`. This matches Ed's story, but
  the trade-off and rubric counts have to work across both models.

## Step 4: Decision variables
Decision variables are the choices Gurobi makes.
Stage 1 choices happen before ratings are known: one plan for every scenario.
Stage 2 choices happen after: a separate plan for each scenario.

```python
# BUILD THE MODEL AND ITS VARIABLES
#   Put Steps 4-6 inside one function of the calculated parameters,
#   so the loop can rebuild the model for each guarantee level.
#
#   pitch[i,d]  yes/no: pitch prospect i at discount d
#               count = prospects x discounts = 90
#   sell[k,s]   whole number: new packages salesperson k sells in scenario s
#               count = salespeople x scenarios = 18
#   fix[k,s]    whole number: makegood fixes salesperson k does in scenario s
#               count = salespeople x scenarios = 18
#
#   Total: 126
#   A pitch isn't a sure sale. It counts as p_close[i,d] of an expected sale.
#
#   If fixes are tracked per client instead:
#   fix[i,s]    count = prospects x scenarios = 90, so the total becomes 198.
```

## Step 5: Constraints
Constraints are the rules every solution must follow. A family is one rule applied
across many prospects, salespeople, or scenarios.

```python
# CONSTRAINTS
#   1. One price per prospect
#      For each prospect: sum over d of pitch[i,d] <= 1
#      count = prospects = 30
#   2. Presale limit (DECIDE which version)
#      sum of pitch <= max_presales, OR sum of p_close x pitch <= max_presales
#      count = 1
#   3. Salesperson hours
#      For each k and s: hours_fix x fix[k,s] + hours_sell x sell[k,s] <= hours[k]
#      count = salespeople x scenarios = 18
#   4. Inventory
#      For each s: sum of p_close x pitch + sum over k of sell[k,s] <= total_packages
#      count = scenarios = 3
#   5. Can't fix clients you don't have
#      For each k and s: fix[k,s] <= sum of p_close x pitch over owned_by[k]
#      count = salespeople x scenarios = 18
#
#   Total: 70 constraints across 5 families.
#
#   If fixes are tracked per client: rule 5 becomes, for each i and s,
#   fix[i,s] <= sum over d of p_close[i,d] x pitch[i,d] (count 90), and rule 3 adds up
#   fixes over owned_by[k]. The total becomes 142.
```

## Step 6: Objectives
Objectives are what we're trying to achieve. Ours conflict.

```python
# OBJECTIVE 1, maximize expected revenue:
#   sum of exp_rev[i,d] x pitch[i,d]
#   + sum over s of prob[s] x stage2_price x sum over k of sell[k,s]
#
# OBJECTIVE 2, minimize expected makegood cost:
#   sum over s of prob[s] x (sum of exp_mg_stage2[i,d,s] x pitch[i,d]
#                            - sum over k of fix_savings[k,s] x fix[k,s])
#   (One-model version. In the two-model version, Model 1 uses exp_mg_stage1.)
#
# Conflict: presales add revenue but also makegood risk,
#           and fixes cut cost but use hours that could have sold packages.
#
# Ed's expected profit (revenue minus makegood cost) isn't a third objective.
# It's one point on our trade-off curve, where both objectives count equally.
```

## Step 7: Trade-off
We can't get the best of both objectives at once, so we map the best possible
combinations for each guarantee level and choose one.

```python
# TRADE-OFF LOOP, for each g in guarantee_levels:
#   1. Run the calculated-parameters function for g.
#   2. Build the model and maximize revenue alone. Record revenue and makegood cost.
#   3. Rebuild and minimize makegood cost alone. Record both.
#      Expect this run to pitch almost no one: no presales means no makegoods.
#      That's the zero-risk end of the curve.
#   4. Pick about 10 makegood budgets between the two costs. For each one, maximize
#      revenue with an added rule: makegood cost <= budget.
#   5. Save each run: g, budget, revenue, makegood cost, who was pitched,
#      and sell/fix by salesperson and scenario.
#   check: if underdelivery is counted on the package total, makegood cost at g = 0
#          is 0 in every run (no shortfall)
```
